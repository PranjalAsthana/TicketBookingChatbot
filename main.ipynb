```python
def search_tickets(origin, destination, max_price=None):

    results = []

    for ticket in tickets:

        if ticket["from"].lower() != origin.lower():
            continue

        if ticket["to"].lower() != destination.lower():
            continue

        if max_price is not None:
            if ticket["price"] > max_price:
                continue

        results.append(ticket)

    return results
```

```python
results = search_tickets(
    "Delhi",
    "Mumbai",
    7000
)

print(results)
```

```python
tools = [
    {
        "type": "function",
        "name": "search_tickets",
        "description": "Search available travel tickets based on origin, destination, and maximum price.",
        "parameters": {
            "type": "object",
            "properties": {
                "origin": {
                    "type": "string",
                    "description": "Departure city"
                },
                "destination": {
                    "type": "string",
                    "description": "Arrival city"
                },
                "max_price": {
                    "type": "number",
                    "description": "Maximum acceptable ticket price"
                }
            },
            "required": [
                "origin",
                "destination"
            ]
        }
    }
]
```

```python
import json

origin = "Delhi"
destination = "Mumbai"
date = "20 September 2026"


response = client.chat_completion(
    messages=[
        {
            "role": "user",
            "content": f"""
You are a ticket booking agent.

You are tasked with booking tickets from {origin}
to {destination} for {date}.

Search for available tickets using the search_tickets tool
and show the user the available options.
Ask the user for confirmation before booking.
"""
        }
    ],
    tools=tools,
    max_tokens=100
)

message = response.choices[0].message

if hasattr(message, "tool_calls") and message.tool_calls:

    tool_call = message.tool_calls[0]

    function_name = tool_call.function.name
    arguments = json.loads(tool_call.function.arguments)

    if function_name == "search_tickets":

        results = search_tickets(
            arguments["origin"],
            arguments["destination"],
            arguments.get("max_price")
        )

        # print("\nAvailable tickets:")
        # print(json.dumps(results, indent=2))
        print("\nAvailable tickets:\n")

        for ticket in results:
            print(f"Flight: {ticket['id']}")
            print(f"Route: {ticket['from']} → {ticket['to']}")
            print(f"Departure: {ticket['departure']}")
            print(f"Arrival: {ticket['arrival']}")
            print(f"Price: ₹{ticket['price']}")
            print("-" * 30)

else:
    print(message.content)
```
